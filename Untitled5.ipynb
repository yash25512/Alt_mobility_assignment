```python
# Create SQLite Database in Memory

# Create an in-memory SQLite database
conn = sqlite3.connect(':memory:')

# Save DataFrames to SQL tables
customer_orders_df.to_sql('customer_orders', conn, index=False, if_exists='replace')
payments_df.to_sql('payments', conn, index=False, if_exists='replace')
```

```python
# Running sql queries
# 1 Order and Sales Analysis

query ='''
SELECT
    order_status,
    COUNT(*) AS total_orders
FROM customer_orders
GROUP BY order_status;
 '''
result = pd.read_sql_query(query, conn)
print(result)
```

```python
# Monthly sales trend (revenue per month)

query1 ='''
          SELECT
    STRFTIME('%Y-%m', order_date) AS order_month,
    SUM(order_amount) AS total_revenue
FROM customer_orders
GROUP BY order_month
ORDER BY order_month;

      '''
result = pd.read_sql_query(query1, conn)
print(result)
```

```python
# Top 5 Customers by Total Spend
query2 ='''
        SELECT
    customer_id,
    SUM(order_amount) AS total_spent
FROM customer_orders
GROUP BY customer_id
ORDER BY total_spent DESC
LIMIT 5;
         '''
result = pd.read_sql_query(query2, conn)
print(result)
```

```python
# Number of Orders per Month

query3='''
        SELECT
    STRFTIME('%Y-%m', order_date) AS order_month,
    COUNT(order_id) AS num_orders
FROM customer_orders
GROUP BY order_month
ORDER BY order_month;

     '''
result = pd.read_sql_query(query3, conn)
print(result)
```

```python
# Customer Analysis Queries
```

```python
#  1. Customers with Repeat Orders

query4='''
  SELECT
    customer_id,
    COUNT(order_id) AS total_orders
FROM customer_orders
GROUP BY customer_id
HAVING total_orders > 1
ORDER BY total_orders DESC;

'''
result = pd.read_sql_query(query4, conn)
print(result)
```

```python
# 2. Total Unique vs. Repeat Customers

query5 ='''
WITH customer_counts AS (
    SELECT customer_id, COUNT(*) AS total_orders
    FROM customer_orders
    GROUP BY customer_id
)
SELECT
    SUM(CASE WHEN total_orders = 1 THEN 1 ELSE 0 END) AS one_time_customers,
    SUM(CASE WHEN total_orders > 1 THEN 1 ELSE 0 END) AS repeat_customers
FROM customer_counts;
'''
result = pd.read_sql_query(query5, conn)
print(result)
```

```python
# 3. Monthly New Customer Trend

query6='''
 WITH first_orders AS (
    SELECT customer_id, MIN(order_date) AS first_order_date
    FROM customer_orders
    GROUP BY customer_id
)
SELECT
    STRFTIME('%Y-%m', first_order_date) AS first_order_month,
    COUNT(customer_id) AS new_customers
FROM first_orders
GROUP BY first_order_month
ORDER BY first_order_month;

'''
result = pd.read_sql_query(query6, conn)
print(result)
```

```python
# 4. Monthly Repeat Order Trend

query7='''
 WITH order_rankings AS (
    SELECT
        customer_id,
        order_id,
        order_date,
        ROW_NUMBER() OVER (PARTITION BY customer_id ORDER BY order_date) AS order_rank
    FROM customer_orders
)
SELECT
    STRFTIME('%Y-%m', order_date) AS order_month,
    COUNT(*) AS repeat_orders
FROM order_rankings
WHERE order_rank > 1
GROUP BY order_month
ORDER BY order_month;
'''
result = pd.read_sql_query(query7, conn)
print(result)
```

```python
#10. Success Rate of Payments
query9='''
SELECT
    ROUND(
        100.0 * SUM(CASE WHEN payment_status = 'completed' THEN 1 ELSE 0 END) / COUNT(*), 2
    ) AS success_percentage
FROM payments;
'''
result = pd.read_sql_query(query9, conn)
print(result)
```

```python
# 11.Failed Payments by Month

query10 ='''
SELECT
    payment_method,
    COUNT(*) AS failed_count
FROM payments
WHERE payment_status = 'failed'
GROUP BY payment_method
ORDER BY failed_count DESC;
'''
result = pd.read_sql_query(query10, conn)
print(result)
```

```python
# 12. . Total Failed Payment Amount
query11='''
SELECT
    SUM(payment_amount) AS total_failed_payment_amount
FROM payments
WHERE payment_status = 'failed';

'''
result = pd.read_sql_query(query11, conn)
print(result)
```

```python
# 13 . Join Orders with Failed Payments (Customer Impact)
query12='''
 SELECT
    co.order_id,
    co.customer_id,
    p.payment_method,
    p.payment_date,
    p.payment_status
FROM payments p
JOIN customer_orders co ON p.order_id = co.order_id
WHERE p.payment_status = 'failed';

'''
result = pd.read_sql_query(query12, conn)
print(result)
```

```python
import pandas as pd
import numpy as np

# Ensure the order_date is in datetime format
orders_df['order_date'] = pd.to_datetime(orders_df['order_date'])

# Extract the year-month from the order date
orders_df['order_month'] = orders_df['order_date'].dt.to_period('M')

# Find the first order month for each customer (cohort)
orders_df['cohort_month'] = orders_df.groupby('customer_id')['order_month'].transform('min')

# Create a column for the month number (e.g., 0 for the first month, 1 for the next, etc.)
orders_df['cohort_index'] = (orders_df['order_month'].dt.year - orders_df['cohort_month'].dt.year) * 12 + (orders_df['order_month'].dt.month - orders_df['cohort_month'].dt.month)

# Now we group by the cohort_month and cohort_index to get the number of customers in each cohort by month
cohort_data = orders_df.groupby(['cohort_month', 'cohort_index'])['customer_id'].nunique().reset_index()

# Pivot the data to have cohort_month as rows and cohort_index as columns
cohort_pivot = cohort_data.pivot(index='cohort_month', columns='cohort_index', values='customer_id')

# Fill NaN with 0s for months that have no data
cohort_pivot = cohort_pivot.fillna(0)

# Calculate retention by dividing each month's customers by the first month (cohort month)
cohort_retention = cohort_pivot.divide(cohort_pivot[0], axis=0)

# Optional: Format the retention matrix as percentages
cohort_retention_percentage = cohort_retention * 100

# Display the retention matrix
cohort_retention_percentage
```

```python
# Save the cohort retention to a CSV file
cohort_retention_percentage.to_csv('/content/cohort.csv', index=True)
```
